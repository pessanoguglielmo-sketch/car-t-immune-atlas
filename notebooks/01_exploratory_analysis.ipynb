# CAR-T Cell Atlas — Exploratory Analysis

Loads the final checkpoint (`data/processed/adata_trajectory.h5ad`) produced by
`run_pipeline.py` and provides an interactive space to explore clusters, marker
genes, differential expression results, and the pseudotime trajectory.

Run `python run_pipeline.py --synthetic` (or without `--synthetic` once you have
real GEO data downloaded) before using this notebook.

```python
import sys
sys.path.insert(0, "../src")

import scanpy as sc
import pandas as pd
from utils import load_checkpoint, MARKER_GENES

sc.settings.set_figure_params(dpi=100, facecolor="white")
adata = load_checkpoint("adata_trajectory")
adata
```

## Cluster / cell-type overview

```python
sc.pl.umap(adata, color=["leiden", "cell_type"], wspace=0.4)
```

```python
adata.obs["cell_type"].value_counts()
```

## Marker gene expression by cluster

```python
all_markers = [g for genes in MARKER_GENES.values() for g in genes if g in adata.var_names]
sc.pl.dotplot(adata, all_markers, groupby="cell_type", standard_scale="var")
```

## Differential expression: Infusion Product vs. Post-infusion blood

```python
from utils import DATA_PROCESSED
de_ip = pd.read_csv(DATA_PROCESSED / "de_ip_vs_blood.csv")
de_ip.sort_values("logfoldchange", ascending=False).head(15)
```

## Pseudotime trajectory

```python
sc.pl.umap(adata, color=["dpt_pseudotime", "cell_type"], wspace=0.4)
```

```python
adata.obs.groupby("cell_type")["dpt_pseudotime"].describe()
```
